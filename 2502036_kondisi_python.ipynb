```python
#Kondisi if adalah kondisi yang akan dieksekusi oleh program jika bernilai benar atau TRUE

nilai = 9

#jika kondisi benar/TRUE maka program akan mengeksekusi perintah dibawahnya

if(nilai > 7):
    print("Sembilan Lebih Besar Dari Angka Tujuh") # Kondisi Benar, Dieksekusi
    
#jika kondisi salah/FALSE maka program tidak akan mengeksekusi perintah dibawahnya

if(nilai > 10):
    print("Sembilan Lebih Besar Dari Angka Sepuluh") # Kondisi Salah, Maka tidak tereksekusi
```

```python
#Kondisi if else adalah jika kondisi bernilai TRUE maka akan dieksekusi pada if, tetapi jika bernilai FALSE maka akan dieksekusi kode pada else

nilai = 3

#Jika pernyataan pada if bernilai TRUE maka if akan dieksekusi, tetapi jika FALSE kode pada else yang akan dieksekusi.

if(nilai > 7):
    print("Selamat Anda Lulus")
else:
    print("Maaf Anda Tidak Lulus")
```

```python
# 1. Definisikan data baris Lia

status_lia = {
    'JANUARI': 'BUKA BLOKIR',
    'FEBRUARI': 'BLOKIR',
    'MARET': 'BLOKIR',
    'APRIL': 2400000,
    'MEI': '',
    'JUNI': '',
    'JULI': ''
}

# 2. Perulangan dengan logika if-elif-else yang telah diperbaiki
for bulan, nilai in status_lia.items():
    
    if nilai == 'BUKA BLOKIR':
        print(f"Bulan {bulan}: Akses telah dibuka.")
        
    elif nilai == 'BLOKIR':
        print(f"Bulan {bulan}: Akses masih diblokir.")
        
    elif isinstance(nilai, (int, float)):
        # Format nominal uang ke gaya Indonesia (contoh: 2.400.000)
        nominal_id = f"{nilai:,}".replace(",", ".")
        print(f"Bulan {bulan}: Terjadi transaksi pencairan sebesar Rp {nominal_id}")
        
    else:
        # Menangani data kosong sesuai instruksi Anda
        print(f"Bulan {bulan}: belum ada pencairan")
```

```python
# 1. Representasi data pencairan kip baris Lia

status_lia = {
    'JANUARI': 'BUKA BLOKIR',
    'FEBRUARI': 'BLOKIR',
    'MARET': 'BLOKIR',
    'APRIL': 2400000, # Data dari pencairan kumulatif sebelumnya
    'MEI': '',
    'JUNI': '',
    'JULI': ''
}

# 2. Konstanta tarif berdasarkan catatan algoritma di Excel
TARIF_BUKA_BLOKIR = 800000

# 3. Variabel penghitung logika algoritma
jumlah_buka_blokir = 0
total_pencairan = 0

print("--- ANALISIS STATUS BULANAN ---")

# 4. Jalankan perulangan logika IF-ELIF-ELSE
for bulan, nilai in status_lia.items():
    
    if nilai == 'BUKA BLOKIR':
        jumlah_buka_blokir += 1
        print(f"Bulan {bulan}: Akses telah dibuka.")
        
    elif nilai == 'BLOKIR':
        print(f"Bulan {bulan}: Akses masih diblokir.")
        
    elif isinstance(nilai, (int, float)):
        # Algoritma menghitung berdasarkan input langsung nominal di Excel
        total_pencairan += nilai
        nominal_id = f"{nilai:,}".replace(",", ".")
        print(f"Bulan {bulan}: Terjadi pencairan sebesar Rp {nominal_id}")
        
    else:
        print(f"Bulan {bulan}: belum ada pencairan")

# 5. Output hitungan algoritma total dana yang cair
print("-" * 50)
total_id = f"{total_pencairan:,}".replace(",", ".")
print(f"Total Bulan Buka Blokir : {jumlah_buka_blokir} bulan")
print(f"TOTAL DANA YANG CAIR   : Rp {total_id}")
```
